```python
new_df['Individual_fare']=new_df['Fare']/(new_df['SibSp']+new_df['Parch']+1)
```

```python
new_df['Family_size']=new_df['Parch']+new_df['SibSp']+1
```

```python
new_df['Family_size'].value_counts()
```

```python
def family_size(num):
    if num == 1:
        return 'alone'
    elif 2<=num<=4:
        return 'small'
    else:
        return 'large'
```

```python
new_df['Family_type']=new_df['Family_size'].apply(family_size)
```

```python
new_df['Family_type'].value_counts()
```

```python
pd.crosstab(new_df['Survived'], new_df['Family_type'],normalize='columns')*100
```

```python
new_df['surname'] = new_df['Name'].str.split(',').str.get(0)
```

```python
new_df['Title']=new_df['Name'].str.split(',').str.get(1).str.strip().str.split(' .').str.get(0)
```

```python
def replace_title(title):
    if title not in ['Mrs.', 'Miss.', 'Mr.', 'Master.']:
        return 'Other'
    return title

new_df['Title'].str.capitalize().str.strip()
new_df['Title']=new_df['Title'].apply(replace_title)
new_df['Title'].value_counts()
```

# LOGISTIC REGRESSION

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.preprocessing import StandardScaler
# Prepare the data
X = new_df_train.drop(columns=['Survived'])
y = new_df_train['Survived']

# Convert categorical variables to dummy variables
X = pd.get_dummies(X, drop_first=True)
new_df_test = pd.get_dummies(new_df_test, drop_first=True)

# Ensure the test set has the same columns as the training set
new_df_test = new_df_test.reindex(columns=X.columns, fill_value=0)

# Split the data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize the scaler
scaler = StandardScaler()

# Fit and transform the training data, and only transform the validation data
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)

# Initialize and train the logistic regression model
logreg = LogisticRegression(max_iter=5000)
logreg.fit(X_train, y_train)

# Predict on the validation set
y_pred = logreg.predict(X_val)

# Evaluate the model
accuracy = accuracy_score(y_val, y_pred)
conf_matrix = confusion_matrix(y_val, y_pred)
class_report = classification_report(y_val, y_pred)

print(f'Accuracy: {accuracy}')
print('Confusion Matrix:')
print(conf_matrix)
print('Classification Report:')
print(class_report)
```

```python
# Predict the survival for the new test data
test_predictions = logreg.predict(new_df_test)
print("Predictions:", test_predictions)

prediction_probabilities = logreg.predict_proba(new_df_test)
print("Prediction Probabilities:", prediction_probabilities)

df=pd.DataFrame(test_predictions,columns=['Survival'])
```
